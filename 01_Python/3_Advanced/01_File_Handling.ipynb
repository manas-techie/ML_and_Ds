### File mode quick refernce
- 'r ' - read only
- 'w ' - write only (truncates file if it exists)
- 'a ' - append (write only, creates file if it doesn't exist)      
- 'r+' - read and write (does not truncate file)
- 'w+' - read and write (truncates file if it exists)
- 'a+' - read and append (creates file if it doesn't exist)

#### using with statement


```python
# The file is automatically closed when the indented block ends
with open(txt_file, "r") as  file:
    content = file.read()
    print(content)
```

```python
# The file is automatically closed when the indented block ends
with open("example.txt", "w") as file:
    file.write("Hello, World!")
```

```python
## two  method to write in a file 

lines = [
    "First line\n",
    "second line\n",
    "third line\n"
]

### its overwrite the files data 
with open("example.txt",'w') as file:
    # writting a single string 
    file.write("Header: My file\n")

    ##writting a list of string
    file.writelines(lines)
```

#### work with csv files

```python
parsed_data=[] 

with open(csv_file, 'r') as file:
    ## read the first line to get the header
    headerline = file.readline()
    ## remove white space and next line and split by comma
    header =headerline.strip().split(",")


    # Iterate over the remaining lines directly
    for line in file:
        # Clean and split the current line
        row_values = line.strip().split(',')
        
        # Combine headers and values into a dictionary using zip()
        row_dict = dict(zip(header, row_values))
        parsed_data.append(row_dict)

for employee in parsed_data:
    print(f"{employee["name"]} works in {employee["city"]}")
```
