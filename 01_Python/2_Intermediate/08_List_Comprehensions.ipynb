```python
# basic list comprehensions

squares = [x ** 2 for x in range(10)]
print(squares)
```

```python
# convert a list of string to uppercase
words = ["hello","world","python"]
uppercase_words = [word.upper() for word in words]
print(uppercase_words)
```

```python
## Condition  list comprehension

even_number = [x for x in range(1,10) if x % 2==0]
print(even_number)
```

```python
## even and odd both using ternary
labels = ['Even' if x % 2 == 0 else 'Odd' for x in range(1,10)]
print(labels)
```

```python
# Nested list comprehension
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

flatted_matrix = [num for row in matrix for num in row]
print(flatted_matrix)
```

```python
## transpose  matrix swap  row and colums
transpose = [[row[i] for row in matrix] for i in range(len(matrix))]
print(transpose)
```

```python
# Dictionary Comprehension
cube = {x: x ** 3 for x in range(1, 11)}
print(cube)
```

```python
## swap key and value in dict

reverse_dict = {value:  key for key , value in cube.items()}
print(reverse_dict)
```

```python
## set comprehension
word  = "kjahnfiuhkfofhskld"
vowel = {x for x in word if x in "aeiouAEIOU"}
print(vowel)
```

```python
### create a set of  string length ignore the duplicate length
names = {"Manas","Alice","Sam","Santanu","Suman","Garg"}
len_set = {len(name) for name in names}
print(len_set)
```
