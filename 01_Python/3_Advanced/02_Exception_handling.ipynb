#### An exception is an event that occurs during the execution of a program that disrupts the normal flow of instructions. When a Python script encounters a situation it cannot cope with, it raises an exception. If not handled, the program will crash. Exception handling allows your program to deal with these errors gracefully and continue running

### Try and Except Block    

```python
## try : block contain the code, which might cause exception while executing
## except:block conatin the code, which  will execute if exception occures in the try block 

def divide_number(a,b):
    try:
        result= a/b
        print(result)
    except ZeroDivisionError:
        # This block run only if ZeroDivisorError occur
        print("Error: you can't divide a number by zero")

divide_number(25,5)

divide_number(10,0)
```

### multiple except clause

```python
def process_user_input(user_input):
    try:
        number = int (user_input)
        result = 10 / number
        print(result)
    except ValueError: 
        print("Error: Please enter a valid Integer")
    except ZeroDivisionError:
        print("Error: you can't divide a number by zero")
    except Exception as e: #catch other unexpected error
        print(e)

process_user_input("5")    # Works perfectly
process_user_input("abc")  # Triggers ValueError
process_user_input("0")
```

### Else clause and Finally clause

```python
### else: Execute only if no exception were raised in the try block
### finally: Execute no matter what happns - This is typically used for the cleanup actions, closing file or raising network connection

def read_file(file_name):
    try:
        file = open(file_name, 'r')
        content = file.read()
    except FileNotFoundError:
        print("This file does not exist")
    else:
        print(content)
    finally:
        #It checks whether a local variable named "file" exists and that its file-like object is not closed.
        if 'file' in locals() and not file.closed:
            file.close()
            print("File Closed")

read_file('sample.txt')
```

### raise keyword - raising an exception

```python
## raise - mannually raise a exception when needed

def is_adult(age):
    if age < 0:
        raise ValueError("Age can't be negative")
    elif age < 18:
        print("minor")
    else:
        print("adult")

# is_adult(-5) -> this will cause a exception and crash the program

try:
    is_adult(25)
    is_adult(-5)
except Exception as e:
    print(e)
```

### Custom Exceptions

#### We can also create our own custom exceptions by defining a new class that inherits from the built-in Exception class. This allows us to create specific error types that can be raised and handled in our code, providing more clarity and control over error handling.

```python
class InsufficientFundsError(Exception):
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        self.message = f"Withdraw amount: {amount} > Balance Amount: {balance}"
        super().__init__(self.message)

# Use the custom exception in a function
def withdraw_money(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(balance, amount)
    
    new_balance = balance - amount
    print(f"Successfully withdrew ${amount}. New balance: ${new_balance}")
    return new_balance

# Handle the custom exception
try:
    current_balance = 50
    withdraw_money(current_balance, 100)
except InsufficientFundsError as e:
    print(f"Transaction Failed: {e}")
```
